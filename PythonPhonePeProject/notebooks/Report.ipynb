# Task 3: Data Quality Checks

## 3.1: Ensure data consistency across state and district levels

### For each state, calculate the total number of transactions, total transaction amount, and total registered users by summing up the values from the district level data.

```python
dist_table = District_Txn_and_Users.groupby(["State"]).agg(dist_total_Transactions = ("Transactions","sum"), dist_total_transaction_amount = ("Amount (INR)", "sum"), dist_total_registered_users = ("Registered Users", "sum")).reset_index().sort_values(["dist_total_Transactions", "dist_total_transaction_amount", "dist_total_registered_users"], ascending = [False, False, False])
dist_table
```

```python
dist_table = dist_table.iloc[0:20, :]
plt.figure()
plt.pie(dist_table["dist_total_transaction_amount"], labels = dist_table["State"], autopct = "%1.1f%%")
plt.show()
```

### Compare the results with the corresponding values at the state level to ensure they match.

```python
state_table = State_Txn_and_Users.groupby(["State"]).agg(state_total_Transactions = ("Transactions","sum"), state_total_transaction_amount = ("Amount (INR)", "sum"), state_total_registered_users = ("Registered Users", "sum"))
state_table
```

### Display any discrepancies found between the district-level and state-level data.

```python
merged = pd.merge(dist_table, state_table, on = "State", how = "inner")

discrepancies = merged[
(merged["dist_total_Transactions"] != merged["state_total_Transactions"]) |
(merged["dist_total_transaction_amount"] != merged["state_total_transaction_amount"]) |
(merged["dist_total_registered_users"] != merged["state_total_registered_users"])
]
discrepancies
```

# Task 4: Data Merging and Advanced Analysis

## 4.1: Ratio of users to population by state

### Merge the State_Txn and Users dataset with the District Demographics dataset to calculate the ratio of registered users to the population for each state. Display the results in a tabular format.

```python
state_tbl = State_Txn_and_Users.groupby(["State"]).agg(RegisteredUsers = ("Registered Users", "sum")).reset_index()
dist_tbl = District_Demographics.groupby(["State"]).agg(Population = ("Population","sum")).reset_index()

fin_tbl = pd.merge(state_tbl,dist_tbl, on = "State", how = "inner" )
fin_tbl["Ratio"] = (fin_tbl["RegisteredUsers"]/fin_tbl["Population"])
fin_tbl.sort_values(["Ratio"], ascending  = True, inplace = True)
fin_tbl
```

### Create a column chart depicting the ratio of users to population by state.

```python
plt.figure()
plt.bar(fin_tbl["State"], fin_tbl["Ratio"])
plt.xticks(rotation = 90)
plt.show()
```

## 4.3: Average transaction amount per user

### Merge relevant datasets to calculate the average transaction amount per user for each state. Display the results in a tabular format.

```python
fin_tbl = (
    State_Txn_and_Users
    .groupby("State")
    .agg(
        TotalTransactionAmount=("Amount (INR)", "sum"),
        TotalUsers=("Registered Users", "sum")
    )
    .reset_index()
)

fin_tbl["AvgTxnPerUser"] = (
    fin_tbl["TotalTransactionAmount"] / fin_tbl["TotalUsers"]
)

fin_tbl
```

# Task 5: Data Visualization

## 5.1: Plot the total transactions and amount over time for a selected state

### Create a line plot showing the total number of transactions and the total transaction amount over time (years and quarters) for a selected state.
[Hint: you can select any state, maybe your home state or state with max transactions]

```python
tbl = State_Txn_and_Users.groupby(["Year","Quarter"]).agg(TotalTransactions = ("Transactions", "sum") , TotalTrxAmt = ("Amount (INR)","sum")).reset_index()
tbl["YearQuarter"] = tbl["Year"].astype(str) + " -Q" + tbl["Quarter"].astype(str)
tbl["txn_norm"] = (tbl["TotalTransactions"] - tbl["TotalTransactions"].min())/(tbl["TotalTransactions"].max() - tbl["TotalTransactions"].min())
tbl["txnamt_norm"] = (tbl["TotalTrxAmt"] - tbl["TotalTrxAmt"].min())/(tbl["TotalTrxAmt"].max() - tbl["TotalTrxAmt"].min())
tbl
plt.figure()
plt.plot(tbl["YearQuarter"], tbl["txn_norm"], marker = "o", label="Transactions")
plt.plot(tbl["YearQuarter"], tbl["txnamt_norm"], marker = "x", label="Transaction Amount")
plt.xticks(rotation = 90)
plt.legend()
plt.grid()
plt.show()
```
